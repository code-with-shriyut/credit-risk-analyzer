# 🏦 Credit Risk Analyzer — EDA
## Home Credit Default Risk Dataset
- **Goal:** Understand the dataset, find patterns, prepare for ML modeling
- **Dataset:** Home Credit Default Risk (Kaggle)
- **Primary File:** application_train.csv (307,511 rows × 122 columns)
- **Target Variable:** TARGET (1 = Defaulted, 0 = Repaid)

```python
# Libraries Import

# Data manipulation
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', 100)
pd.set_option('display.float_format', '{:.2f}'.format)

print("✅ Libraries imported successfully!")
```

```python
# Load the dataset
df = pd.read_csv('../data/raw/application_train.csv') # pd.read_csv() is used to read a CSV file and create a DataFrame from it. The path to the CSV file is provided as an argument.
print(f"Shape: {df.shape}")
print(f"Rows: {df.shape[0]:,}") # Print the number of rows with a thousands separator
print(f"Columns: {df.shape[1]}") # Print the number of columns
```

```python
# Preview first 5 rows of the dataset
df.head(5)
```

```python
# Data types and non-null counts
df.info()
```

```python
#Target Variable Analysis

# Target distribution
print("Target Value Counts:")
print(df['TARGET'].value_counts())
print()
print("Target Percentage:")
print(df['TARGET'].value_counts(normalize=True).mul(100).round(2)) # value_counts(normalize=True) gives the relative frequencies of each unique value in the 'TARGET' column, mul(100) converts it to percentage, and round(2) rounds the percentages to 2 decimal places.

# Visualize
plt.figure(figsize=(6, 4))
df['TARGET'].value_counts().plot(kind='bar', color=['#2196F3', '#F44336'], edgecolor='black')
plt.title('Target Variable Distribution')
plt.xlabel('TARGET (0 = Repaid, 1 = Defaulted)')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### 📊 Insights — Target Variable Distribution

- Dataset is **heavily imbalanced** — 92% loans were repaid, only 8% defaulted
- If we train model directly, it will just predict 0 for everyone and still show 92% accuracy — this is called the **accuracy paradox**
- We will use `scale_pos_weight = 11.4` in XGBoost to fix this
- We will use **ROC-AUC** instead of accuracy as our evaluation metric
- This is normal in real banking data — defaults are always a small minority

```python
# Missing Values Analysis

missing = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_percent}).sort_values('Missing %', ascending = False) # Create a DataFrame to display missing counts and percentages, and sort it by percentage in descending order.
# Only show columns with missing values
missing_df = missing_df[missing_df['Missing Count'] > 0] # Filter the DataFrame to include only columns that have missing values (i.e., where 'Missing Count' is greater than 0).

print(f"Columns with Missing Values: {len(missing_df)}") # Print the number of columns that have missing values.
print() # Print a blank line for better readability
print(missing_df.head(20)) # Display the top 20 columns with missing values.
```

### 📊 Insights — Missing Values Analysis

- **67 out of 122 columns** have missing values — more than half the dataset!
- Most missing columns are **building/apartment related** — many applicants didn't provide home details
- `OWN_CAR_AGE` is 66% missing — because those applicants simply **don't own a car**
- `EXT_SOURCE_1` is 56% missing — very important feature, needs careful handling
- `EXT_SOURCE_2` is almost complete (0.21%) — most reliable external credit score

```python
# Check EXT_SOURCE missing values specifically
ext_cols = ['EXT_SOURCE_1', 'EXT_SOURCE_2', 'EXT_SOURCE_3']
print("EXT_SOURCE Missing Values:")
for col in ext_cols:
    missing_pct = df[col].isnull().sum() / len(df) * 100
    print(f"{col}: {missing_pct:.2f}% missing")
```

```python
# Check for the 365243 value in DAYS_EMPLOYED

print("Value 365243 count:", (df['DAYS_EMPLOYED'] == 365243).sum())
print("Percentage:", round((df['DAYS_EMPLOYED'] == 365243).sum() / len(df) * 100, 2), "%")
```

```python
# Visualize the anomaly in DAYS_EMPLOYED, the 365243 value is an outlier and likely represents a placeholder for missing data, as it corresponds to approximately 1000 years of employment, which is unrealistic. To visualize this anomaly, we can create histograms of the 'DAYS_EMPLOYED' column before and after replacing the 365243 value with NaN.

fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Before fix
axes[0].hist(df['DAYS_EMPLOYED'].dropna(), bins=50, 
             color='#F44336', edgecolor='black', alpha=0.7)
axes[0].set_title('DAYS_EMPLOYED — Before Fix', fontweight='bold')
axes[0].set_xlabel('Days')
axes[0].set_ylabel('Count')

# After fix
clean = df['DAYS_EMPLOYED'].replace(365243, np.nan) # Replace the 365243 value with NaN to clean the data for visulization. This creates a new Series called 'clean' where all occurrences of 365243 in the 'DAYS_EMPLOYED' column are replaced with NaN, allowing us to visualize the distribution without the outlier skewing the results.
axes[1].hist(clean.dropna(), bins=50, 
             color='#4CAF50', edgecolor='black', alpha=0.7)
axes[1].set_title('DAYS_EMPLOYED — After Fix', fontweight='bold')
axes[1].set_xlabel('Days')

plt.suptitle('DAYS_EMPLOYED Anomaly — Before vs After', fontweight='bold')
plt.tight_layout()
plt.show()
```

### 📊 Insights — DAYS_EMPLOYED Anomaly Fix

- **55,374 applicants (18%)** had the placeholder value 365243
- After replacing with NaN, the distribution looks **clean and realistic**
- Most applicants have been employed for **0 to 3000 days (0 to 8 years)**
- These applicants (18%) are likely **unemployed, retired, or self-employed**
- In preprocessing, we will permanently replace 365243 with NaN in the main dataframe
